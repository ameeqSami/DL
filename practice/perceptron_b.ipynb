```python
import numpy as np 
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron
```

```python
iris = load_iris()
```

```python
iris.feature_names
```

```python
iris.target_names
```

```python
iris.target
```

```python
X = iris.data[:, (2, 3)]
y = (iris.target == 0).astype(int) # Iris setosa?
per_clf = Perceptron()
per_clf.fit(X, y)
y_pred = per_clf.predict([[2, 0.5]])
```
